# 02 - Construcción de la Homografía

Selección de puntos correspondientes entre la imagen de la cancha y las coordenadas reales, cálculo de la matriz H y guardado.

```python
import cv2
import numpy as np
```

```python
# Cargar imagen de referencia de la cancha
img = cv2.imread("../data/raw/frame_referencia.png")
print("Dimensiones:", img.shape)
```

```python
# Puntos en la imagen (píxeles) - ajustar según la imagen real
pts_src = np.array([
    [300, 400],
    [900, 400],
    [900, 700],
    [300, 700]
], dtype=np.float32)

# Puntos reales en la cancha (metros) - esquinas conocidas
pts_dst = np.array([
    [0.0,  0.0],
    [68.0, 0.0],
    [68.0, 52.0],
    [0.0,  52.0]
], dtype=np.float32)
```

```python
# Calcular la homografía
H, status = cv2.findHomography(pts_src, pts_dst)
print("Matriz H:\n", H)
print("Status:", status.ravel())
```

```python
# Guardar la matriz H
np.save("../models/homographies/H_matrix.npy", H)
print("Homografía guardada en models/homographies/H_matrix.npy")
```
